```python
@jit
def f_implicit(a):
    return (a*b.shape[0]).sum()

@jit
def f_explicit(a, b):
    return (a*b.shape[0]).sum()
```

```python
import equinox as eqx
```

```python
class ExampleModule(eqx.Module):
    m : jax.Array
    def __init__(self, m):
        self.m = m
    
    @jit    
    def sum_with_m(self, summand):
        return summand + self.m
```

```python
model = ExampleModule(m=jnp.array([1, 2, 3]))
```

```python
@jit
def fun_that_uses_module(number):
    return model.sum_with_m(number**2)
```

```python
fun_that_uses_module(10)
```
